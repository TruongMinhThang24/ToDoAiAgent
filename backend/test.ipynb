```python
import os
from google import genai  # Giữ SDK beta tạm để test
from todo_backend.config.setting import settings  # Import settings của bạn

# Set API key nếu chưa
os.environ["GOOGLE_API_KEY"] = settings.GEMINI_API_KEY  # Hoặc hardcode tạm cho test

client = genai.Client(api_key=settings.GEMINI_API_KEY)

# List all models
models = client.models.list()
print("Available Models:")
for model in models:
    print(f"- {model.name}: Supported generation methods: {model.supported_generation_methods}")
    if "generateContent" in model.supported_generation_methods:
        print(f"  -> SUPPORTS generateContent (good for audio/text)")

# Filter multimodal models
multimodal_models = [m for m in models if "gemini-1.5" in m.name]
print("\nMultimodal Models (for audio):", [m.name for m in multimodal_models])
```
